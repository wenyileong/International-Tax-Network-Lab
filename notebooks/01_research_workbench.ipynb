# International Tax Network Lab: research workbench

This notebook uses the same service layer (`taxnetlab.api`) as the desktop interface, so runs made here are stored in the same workspace and can be replayed from the interface or the command line. The data are synthetic. The three published-method engines are at *interface* status and refuse to run until their specifications are verified.

````python
from taxnetlab.api import Api
from taxnetlab.workspace import Workspace

api = Api(Workspace())  # Documents\\TaxNetLab on Windows, ~/TaxNetLab elsewhere
api.ensure_default_dataset()
[(d['dataset_id'], d['name'], d['data_status']) for d in api.datasets()]
````

````python
did = api.datasets()[0]['dataset_id']
ds = api.ws.load_dataset(did)
print(ds.validate().format_text())
ds.counts()
````

## Views

Ownership edges run holder → held; the value-flow view reverses them. Parallel relations are kept as separate edges (a `networkx.MultiDiGraph`).

````python
from taxnetlab.views import ownership_view, value_flow_view, jurisdiction_matrix

own = ownership_view(ds)
flow = value_flow_view(ds)
own.meta, flow.meta['direction'], own.graph.number_of_edges()
````

````python
m = jurisdiction_matrix(ds, rule='count')
m['rows'], m['values']
````

## A recorded run

The demonstration engine is a pipeline test, not a research method.

````python
out = api.run({'engine_id': 'demo.link_summary', 'dataset_id': did, 'parameters': {},
              'research_question': 'Notebook demonstration'})
rec = out['record']
rec['run_id'], rec['outcome'], rec['outputs']
````

````python
api.replay(rec['run_id'])
````
